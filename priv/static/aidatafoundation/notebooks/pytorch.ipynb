## Prepare the lab files

This cell downloads only the listed lab modules and synthetic data from a pinned source commit, then checks SHA-256 hashes. Inspect the source before running it. Outputs are stored in a temporary runtime directory: download files you want to keep before the runtime ends.


```python
import hashlib
import os
import sys
import tempfile
from pathlib import Path
from urllib.request import urlopen

workshop_root = Path(tempfile.mkdtemp(prefix="ic-pytorch-"))
_workshop_assets = {'labs/common.py': 'f1f8959a4100eb404c0317a5b4db7b377de0af795aece9ffe68e83e8988ec046', 'data/pytorch-regression.pt': '2f588394e1aff3be3ce2ed7f99c67a85cac361700369ebdf172b02adf71e08df', 'data/pytorch-moons.pt': 'b7c0316af60e41ba6a57afb422cfeead706d0c1a9a8ac6bf79f80012ae53f06e', 'labs/lab01_tensors_autograd.py': '5fa96bed8b7c409e69b0656c43e1b96796441053d75366eed1240897e4707022', 'labs/lab02_mlp_classification.py': '5e3d9cf8d0c469a7f117afb150aaaa69fbd0c857f4a1e8010849ef424f50b285'}
for _relative, _expected in _workshop_assets.items():
    _url = "https://raw.githubusercontent.com/inferencecollective/inferencecollective.github.io/a9c6139d8d9ff9e6aae8b7b28d41e7f796639a95/" + _relative
    with urlopen(_url, timeout=30) as _response:
        _payload = _response.read()
    if hashlib.sha256(_payload).hexdigest() != _expected:
        raise RuntimeError("Source verification failed: " + _relative)
    _target = workshop_root / _relative
    _target.parent.mkdir(parents=True, exist_ok=True)
    _target.write_bytes(_payload)
(workshop_root / "labs/__init__.py").touch()
os.environ["MPLBACKEND"] = "Agg"
sys.path.insert(0, str(workshop_root))
# Re-evaluating setup must import the new copies, not cached lab modules.
for _name in list(sys.modules):
    if _name == "labs" or _name.startswith("labs."):
        del sys.modules[_name]
print("Workshop files verified. Outputs:", workshop_root / "outputs")
```
